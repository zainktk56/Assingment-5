**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
# Write your code here for Q2

class BankAccount():
    def __init__(self,name):
        self.name=name
        self.__balance = 0 
        print(f"Account Opened")
        
    def deposit(self,balance):
        self.__balance=balance
        print(f"""
              ******Deposit Successfull ! ******
              
              ******UPDATED BALANCE******
              
              Balance : {self.__balance} 
              Beneficiary Name : {self.name}
              """)
        
    def withdraw(self, withdraw_amt):
        try:  
            
            if withdraw_amt <= 0:
                raise ValueError("Withdrawal amount must be greater than zero.")
                
            if withdraw_amt > self.__balance:
                raise ValueError("Insufficient Balance.")
            
            self.__balance -= withdraw_amt
        
            print(f"""
              ***Withdrawal Successful***
              Withdrawal Amount : {withdraw_amt}
              Remaining Balance : {self.__balance}""")        
                 
        except ValueError as error:
            print(f"Transaction Failed: {error}")
            
           
        
```

```python
p1=BankAccount("ali")
p1.deposit(1000)
p1.withdraw(89)


```

**Q6:** Single Inheritance ka use karte hue aik parent class `Vehicle` (with brand attribute) aur child class `Car` (with model attribute) banayein. Car ka object bana kar dono attributes print karein.

```python
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

class Car(Vehicle):
    def __init__(self, brand, model):
        super().__init__(brand) 
        self.model = model
        
        print(f"CAR MODEL : {self.model}")
        print(f"CAR BRAND : {self.brand}")


```
