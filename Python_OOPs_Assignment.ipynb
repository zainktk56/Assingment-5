**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
# Write your code here for Q2

class BankAccount():
    def __init__(self,name):
        self.name=name
        self.__balance = 0 
        print(f"Account Opened")
        
    def deposit(self,balance):
        self.__balance=balance
        print(f"""
              ******Deposit Successfull ! ******
              
              ******UPDATED BALANCE******
              
              Balance : {self.__balance} 
              Beneficiary Name : {self.name}
              """)
        
    def withdraw(self, withdraw_amt):
        try:  
            
            if withdraw_amt <= 0:
                raise ValueError("Withdrawal amount must be greater than zero.")
                
            if withdraw_amt > self.__balance:
                raise ValueError("Insufficient Balance.")
            
            self.__balance -= withdraw_amt
        
            print(f"""
              ***Withdrawal Successful***
              Withdrawal Amount : {withdraw_amt}
              Remaining Balance : {self.__balance}""")        
                 
        except ValueError as error:
            print(f"Transaction Failed: {error}")
            
           
        
```

```python
p1=BankAccount("ali")
p1.deposit(1000)
p1.withdraw(89)


```

**Q11:** Method Overriding kya hai? Parent class `Bird` banayein jisme `fly()` method ho jo print kare 'Most birds can fly'. Child class `Ostrich` banayein jo is method ko override kare aur print kare 'Ostriches cannot fly'.

```python
# Write your code here for Q11

# Method Over riding also Known as Polymorphism.

class Bird():
    def __init__(self):
        pass
    def fly(self):
        print("Most Birds Cannot fly.")
class Ostrich(Bird):
    
    def __init__(self):
        super().__init__()
    def fly(self):
        print("Ostriches Cannot Fly.") 
        
```

```python
birds=Bird()
birds.fly()

ostrich=Ostrich()
ostrich.fly()
```

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
# Write your code here for Q13

class Number:
    def __init__(self, amt):
        self.amt = amt
        
    def __add__(self, other):
        return Number(self.amt + other.amt)
          
    def __str__(self):
        return str(self.amt)
          
    
        
```

```python
n1 = Number(89)
n2 = Number(10)

total = n1+ n2
print(total)
```

**Q17:** Q16 wali `Computer` class ko inherit karke do child classes `Laptop` aur `Desktop` banayein aur dono mein `process()` method ko mandatory implement karein.

```python
# Write your code here for Q17

from abc import ABC, abstractmethod
class Computer(ABC):
    @abstractmethod
    def process(self):
        pass
    
class Laptop(Computer):
    
    def process(self):
       return "Laptop"
   
class Desktop(Computer):
    
    def process(self):
        return "Desktop "
   
```

**Q18:** Aik abstract class `Car` banayein jisme aik concrete (normal) method `start_engine()` ho aur aik abstract method `drive()` ho. Child class bana kar dono ko test karein.

```python
# Write your code here for Q18
from abc import ABC, abstractmethod

class Car(ABC):
    def start_engine(self):
        return "Engine Started"
        
    @abstractmethod
    def drive(self):
        pass

class SportsCar(Car):
    def drive(self):
        return "Sports car full speed mein track par daud rahi hai!"

```

```python
my_car = SportsCar()

print(my_car.start_engine())
print(my_car.drive())
```

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
# Write your code here for Q20

from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def process_payment(self, amount):
        pass

class CreditCardPayment(Payment):
    def process_payment(self, amount):
        return f"{amount} rupees Credit Card ke zariye pay ho gaye hain."

class UPIPayment(Payment):
    def process_payment(self, amount):
        return f"{amount} rupees UPI ke zariye pay ho gaye hain."
    
    
```

```python
cc = CreditCardPayment()
upi = UPIPayment()

print(cc.process_payment(1500))
print(upi.process_payment(450))
```
